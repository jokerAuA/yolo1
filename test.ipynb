```python
import torch
import torchvision
print(torch.__version__)           # 应该带 +cu118 或 +cu121 等
print(torchvision.__version__)
print(torch.cuda.is_available())   # 必须是 True

# 验证 NMS 有 CUDA 支持
boxes = torch.tensor([[0,0,1,1],[0,0,0.9,0.9]], dtype=torch.float32).cuda()
scores = torch.tensor([0.9, 0.8]).cuda()
print(torchvision.ops.nms(boxes, scores, 0.5))  # 不报错就成功了
```
